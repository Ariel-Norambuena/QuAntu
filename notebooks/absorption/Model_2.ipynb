# **Theoretical model — Isotopic Doppler absorption spectrum in rubidium**

## **1. Physical picture**

In a rubidium cell without buffer gas, **thermal Doppler broadening** is the dominant spectral-broadening mechanism.

Atomic velocities follow the Maxwell–Boltzmann distribution:

$$
f(v)\propto e^{-mv^2/2k_BT}
$$

where

- $m$ is the atomic mass,
- $k_B$ is the Boltzmann constant,
- $T$ is the temperature.

Thermal motion shifts the optical frequency experienced by the atoms through the Doppler effect:

$$
\omega'=\omega-kv
$$

where

- $\omega$ is the laser frequency,
- $k$ is the wave number,
- $v$ is the atomic velocity.

Different velocity classes absorb at slightly different frequencies, producing Gaussian Doppler profiles.

---

# **2. Isotopic composition of natural rubidium**

Natural rubidium consists mainly of two isotopes:

- $^{85}\mathrm{Rb}$
- $^{87}\mathrm{Rb}$

with approximate natural abundances

$$
\eta_{85}=72.17\%
$$

$$
\eta_{87}=27.83\%
$$

Each isotope has

- a different atomic mass,
- a different hyperfine structure,
- different effective absorption positions.

The total spectrum is therefore modeled as the sum of contributions from both isotopes.

---

# **3. Doppler broadening**

The Doppler width of each isotope is

$$
\Delta \nu_D
=
\nu_0
\sqrt{
\frac{8k_BT\ln2}{mc^2}
}
$$

where

- $\nu_0$ is the optical center frequency,
- $m$ is the mass of the isotope,
- $c$ is the speed of light.

Because

$$
m_{85}\neq m_{87}
$$

each isotope has a slightly different Doppler width.

---

# **4. Relation between FWHM and sigma**

For a Gaussian profile:

$$
\mathrm{FWHM}
=
2\sqrt{2\ln2}\,\sigma
$$

therefore

$$
\sigma
=
\frac{\mathrm{FWHM}}
{2\sqrt{2\ln2}}
$$

---

# **5. Isotopic absorption model**

The total absorption is modeled as

$$
A(\Omega)
=
A_{85}(\Omega)
+
A_{87}(\Omega)
$$

where each isotopic contribution is a sum of Gaussian Doppler profiles.

---

# **6. Contribution of $^{85}\mathrm{Rb}$**

The absorption associated with $^{85}\mathrm{Rb}$ is modeled as

$$
A_{85}(\Omega)
=
\eta_{85}
\sum_i
A_i^{(85)}
\exp
\left[
-\frac{
(\Omega-\Omega_i^{(85)})^2
}{
2\sigma_{85}^2
}
\right]
$$

where

- $\eta_{85}$ is the natural abundance,
- $A_i^{(85)}$ is the relative amplitude,
- $\Omega_i^{(85)}$ is the effective peak position,
- $\sigma_{85}$ is the isotope's Doppler width.

---

# **7. Contribution of $^{87}\mathrm{Rb}$**

The contribution from $^{87}\mathrm{Rb}$ is

$$
A_{87}(\Omega)
=
\eta_{87}
\sum_j
A_j^{(87)}
\exp
\left[
-\frac{
(\Omega-\Omega_j^{(87)})^2
}{
2\sigma_{87}^2
}
\right]
$$

where

- $\eta_{87}$ is the natural abundance,
- $A_j^{(87)}$ is the relative amplitude,
- $\Omega_j^{(87)}$ is the effective peak position,
- $\sigma_{87}$ is the isotope's Doppler width.

---

# **8. Effective hyperfine positions**

Effective theoretical positions are introduced:

$$
\Omega_i^{\mathrm{theory}}
$$

associated with the isotopic hyperfine structure of the rubidium D2 line.

### **Theoretical reference positions**

The initial model positions were selected to be consistent with the known hyperfine structure of the rubidium D2 line and used as references for the subsequent phenomenological fit.

These are effective model parameters and do not necessarily coincide with the exact hyperfine frequencies reported in spectroscopic tables.

For $^{87}\mathrm{Rb}$, the positions are

$$
\Omega_{87}^{\mathrm{theory}}=(-5.2,\;2.2)\ \mathrm{GHz}
$$

while for $^{85}\mathrm{Rb}$ they are

$$
\Omega_{85}^{\mathrm{theory}}=(-1.8,\;1.0)\ \mathrm{GHz}.
$$

### **Phenomenological corrections**

Because of

- finite experimental resolution,
- Doppler overlap between resonances,
- frequency-axis calibration,
- effective mixing of hyperfine transitions,
- instrumental limitations associated with digitization of the experimental figure,

the observed positions do not exactly match the theoretical reference positions.

Phenomenological corrections are therefore introduced:

$$
\Omega_i^{\mathrm{eff}}=\Omega_i^{\mathrm{theory}}+\delta_i
$$

where $\delta_i$ is an adjustable effective shift.

The corrections used are

$$
\delta_{87}=(0.54,\;0.00)\ \mathrm{GHz}
$$

$$
\delta_{85}=(-0.25,\;0.00)\ \mathrm{GHz}.
$$

The final effective positions used in the simulation are

$$
\Omega_{87}^{\mathrm{eff}}=(-4.66,\;2.20)\ \mathrm{GHz}
$$

$$
\Omega_{85}^{\mathrm{eff}}=(-2.05,\;1.00)\ \mathrm{GHz}.
$$

---

# **9. Effective spectral intensities**

Each resonance is weighted by an effective spectral weight

$$
W_i
$$

which controls each transition's relative contribution to the total absorption spectrum.

These dimensionless parameters were fitted phenomenologically to reproduce the observed relative intensities.

For $^{87}\mathrm{Rb}$, the positions are

$$
W_i^{(87)}=(0.33,\;1.15)
$$

while for $^{85}\mathrm{Rb}$ they are

$$
W_i^{(85)}=(0.32,\;1.00).
$$

The final observed peak heights do not depend solely on these spectral weights.

Each isotopic contribution is also weighted by its natural abundance:

$$
\eta_{85}=72.17\%
$$

$$
\eta_{87}=27.83\%.
$$

The effective contribution of each resonance to the total spectrum is therefore

$$
A_i^{\mathrm{eff}}=\eta_k W_i^{(k)}
$$

where

$$
k\in\{85,87\}.
$$

Although some internal spectral weights exceed unity, the final observed peak heights result from combining these weights with the natural isotopic abundances.

The complete spectrum is then normalized to its maximum to facilitate comparison with the experimental data.

---

# **10. Global spectrum shift**

A global shift is also introduced:

$$
\Omega \rightarrow \Omega+\Delta
$$

to correct systematic offsets in the horizontal calibration of the experimental spectrum.

---

# **11. Physical parameters**

## Temperature

The value used is

$$
T=293\ \mathrm{K}
$$

approximately room temperature.

---

## Optical line

The rubidium D2 line is used:

$$
\lambda_0=780.241\ \mathrm{nm}
$$

therefore

$$
\nu_0=\frac{c}{\lambda_0}
$$

---

# **12. Doppler widths**

The Gaussian widths of each isotope are calculated from

$$
\Delta \nu_D
=
\nu_0
\sqrt{
\frac{8k_BT\ln2}{mc^2}
}
$$

and then

$$
\sigma
=
\frac{\mathrm{FWHM}}
{2\sqrt{2\ln2}}
$$

giving

$$
\sigma_{85}\neq\sigma_{87}
$$

because the isotope masses differ.

---

# **13. Physical interpretation**

- Each Gaussian profile represents an effective contribution associated with experimentally observed hyperfine transitions.

- The spectral width arises mainly from thermal Doppler broadening.

- The model explicitly includes the contributions of $^{85}\mathrm{Rb}$ and $^{87}\mathrm{Rb}$.

- Effective experimental positions are corrected through adjustable phenomenological parameters.

- Relative intensities are modeled using corrected effective spectral weights.

- The model reproduces the overall shape of the digitized experimental spectrum.

---

# **14. Model limitations**

The model does not include

- Voigt profiles,
- collisional broadening,
- optical Bloch equations,
- optical saturation,
- Zeeman structure,
- complex susceptibility,
- quantum interference,
- Clebsch–Gordan coefficients,
- density-matrix dynamics.

The model is therefore an isotopic phenomenological approximation motivated by Doppler-broadening physics.

---

### **RUBIDIUM ABSORPTION SPECTRUM**

```python
# ============================================================
# RUBIDIUM ABSORPTION SPECTRUM
#
# Improved isotopic Doppler Gaussian model
#
# Includes:
#   - Rb85 and Rb87 separately
#   - Real atomic masses
#   - Natural isotopic abundance
#   - Global spectral shift
#   - Effective peak position corrections
#   - Effective spectral weight corrections
#
# ============================================================
# PHYSICAL CONSTANTS
# ============================================================

# Speed of light [m/s]
c = 299792458                       # [m s^-1]

# Boltzmann constant [J/K]
kB = 1.380649e-23                  # [J K^-1]

# Atomic mass unit [kg]
amu = 1.66053906660e-27            # [kg]


# ============================================================
# RUBIDIUM ISOTOPE MASSES
# ============================================================

# Rb-85 atomic mass [kg]
m85 = 84.911789 * amu              # [kg]

# Rb-87 atomic mass [kg]
m87 = 86.909183 * amu              # [kg]


# ============================================================
# RUBIDIUM D2 TRANSITION
# ============================================================

# D2 transition wavelength [m]
lambda0 = 780.241e-9               # [m]

# Central optical frequency [Hz]
nu0 = c / lambda0                  # [Hz]


# ============================================================
# TEMPERATURE
# ============================================================

# Vapor temperature [K]
T = 293                            # [K]


# ============================================================
# DOPPLER WIDTH FUNCTION
# ============================================================

def doppler_sigma(mass):

    FWHM_Hz = nu0 * np.sqrt(
        (8 * kB * T * np.log(2)) /
        (mass * c**2)
    )

    FWHM_GHz = FWHM_Hz / 1e9

    sigma = FWHM_GHz / (
        2 * np.sqrt(2 * np.log(2))
    )

    return sigma


# ============================================================
# DOPPLER WIDTHS
# ============================================================

sigma85 = doppler_sigma(m85)       # [GHz]
sigma87 = doppler_sigma(m87)       # [GHz]

print("sigma85 =", sigma85, "[GHz]")
print("sigma87 =", sigma87, "[GHz]")


# ============================================================
# NATURAL ISOTOPIC ABUNDANCE
# ============================================================

ab85 = 0.7217
ab87 = 0.2783


# ============================================================
# LOAD EXPERIMENTAL DATA
# ============================================================

data = np.loadtxt(
    "Data_Rb_Gas.csv",
    delimiter=",",
    skiprows=1
)


# ============================================================
# EXPERIMENTAL DATA
# ============================================================

x_exp = data[:,0]                  # [GHz]
y_exp = data[:,1]                  # [arb. units]

y_exp = y_exp / np.max(y_exp)


# ============================================================
# THEORETICAL HYPERFINE PEAK POSITIONS
# ============================================================

centers87_theory = np.array([
    -5.2,
     2.2
])                                  # [GHz]

centers85_theory = np.array([
    -1.8,
     1.0
])                                  # [GHz]


# ============================================================
# EFFECTIVE POSITION CORRECTIONS
# ============================================================

delta87 = np.array([
    0.54,
    0.00
])                                  # [GHz]

delta85 = np.array([
   -0.25,
    0.00
])                                  # [GHz]


# ============================================================
# EFFECTIVE PEAK POSITIONS
# ============================================================

centers87 = centers87_theory + delta87
centers85 = centers85_theory + delta85


# ============================================================
# THEORETICAL SPECTRAL WEIGHTS
# ============================================================

weights87_theory = np.array([
    0.12,
    0.55
])                                  # [dimensionless]

weights85_theory = np.array([
    0.32,
    1.00
])                                  # [dimensionless]


# ============================================================
# EFFECTIVE SPECTRAL WEIGHT CORRECTIONS
# ============================================================

delta_weights87 = np.array([
    0.21,
    0.6
])                                  # [dimensionless]

delta_weights85 = np.array([
    0.00,
    0.00
])                                  # [dimensionless]


# ============================================================
# EFFECTIVE SPECTRAL WEIGHTS
# ============================================================

weights87 = (
    weights87_theory
    + delta_weights87
)

weights85 = (
    weights85_theory
    + delta_weights85
)


# ============================================================
# GLOBAL SPECTRAL SHIFT
# ============================================================

shift = 0.35                        # [GHz]


# ============================================================
# DETUNING AXIS
# ============================================================

margin = abs(shift)

Omega = np.linspace(
    np.min(x_exp) - margin,
    np.max(x_exp) + margin,
    50000
)                                   # [GHz]


# ============================================================
# BUILD ABSORPTION SPECTRUM
# ============================================================

Absorption = np.zeros_like(Omega)


# ============================================================
# Rb87 CONTRIBUTION
# ============================================================

for W, O0 in zip(weights87, centers87):

    gaussian87 = (
        ab87
        * W
        * np.exp(
            -(Omega - O0)**2 /
            (2 * sigma87**2)
        )
    )

    Absorption += gaussian87


# ============================================================
# Rb85 CONTRIBUTION
# ============================================================

for W, O0 in zip(weights85, centers85):

    gaussian85 = (
        ab85
        * W
        * np.exp(
            -(Omega - O0)**2 /
            (2 * sigma85**2)
        )
    )

    Absorption += gaussian85


# ============================================================
# NORMALIZATION
# ============================================================

Absorption /= np.max(Absorption)


# ============================================================
# APPLY GLOBAL SHIFT
# ============================================================

Omega_shifted = Omega + shift


# ============================================================
# FIGURE
# ============================================================

plt.figure(figsize=(10,6))


# ============================================================
# EXPERIMENTAL DATA
# ============================================================

plt.scatter(
    x_exp,
    y_exp,
    color='black',
    s=25,
    label='Experimental data'
)


# ============================================================
# THEORETICAL MODEL
# ============================================================

plt.plot(
    Omega_shifted,
    Absorption,
    color='red',
    linewidth=3,
    label='Improved isotopic Gaussian model'
)


# ============================================================
# HYPERFINE LABELS
# ============================================================

# ------------------------------------------------------------
# Left peak : Rb87 F=1
# ------------------------------------------------------------

plt.text(
    -4.3,
    0.16,
    r'$^{87}\mathrm{Rb}$' + '\n' + r'$F=1$',
    fontsize=18,
    ha='center'
)


# ------------------------------------------------------------
# Second peak : Rb85 F=2
# ------------------------------------------------------------

plt.text(
    -1.76,
    0.36,
    r'$^{85}\mathrm{Rb}$' + '\n' + r'$F=2$',
    fontsize=18,
    ha='center'
)


# ------------------------------------------------------------
# Main peak : Rb85 F=3
# ------------------------------------------------------------

plt.text(
    0.24,
    0.88,
    r'$^{85}\mathrm{Rb}$' + '\n' + r'$F=3$',
    fontsize=18,
    ha='center'
)


# ------------------------------------------------------------
# Right peak : Rb87 F=2
# ------------------------------------------------------------

plt.text(
    2.6,
    0.48,
    r'$^{87}\mathrm{Rb}$' + '\n' + r'$F=2$',
    fontsize=18,
    ha='center'
)


# ============================================================
# CELL LABEL
# ============================================================

plt.text(
    4.25,
    0.95,
    'Pure Rb',
    fontsize=20
)

# ============================================================
# STYLE
# ============================================================

plt.xlabel(
    'Laser detuning [GHz]',
    fontsize=14
)

plt.ylabel(
    'Normalized absorption [a.u.]',
    fontsize=14
)

# plt.title(
#     f'Rubidium absorption spectrum (T = {T} K)',
#     fontsize=18
# )

plt.grid(alpha=0.3)

plt.legend(
    fontsize=12,
    loc='upper left'
)

plt.tight_layout()

plt.show()
```

### **Difference between experiment and theory**

The lower panel shows the difference between the experimental spectrum and the fitted Gaussian Doppler model.

The theoretical spectrum is interpolated onto the experimental grid, after which the residual is calculated:

$$
R(\Omega)=A_{\mathrm{exp}}-A_{\mathrm{model}}
$$

The difference is multiplied by 100 to highlight small discrepancies between theory and experiment.

Values close to zero indicate good agreement between the model and experimental data.

# **Fit-quality metrics**

The mean squared error (MSE) and mean absolute error (MAE) quantify agreement between the theoretical model and experimental data.

Let

$$
y_i
$$

denote an experimental value and

$$
\hat y(x_i)
$$

the model prediction evaluated at the same point.

The MSE is defined as

$$
\mathrm{MSE}=

\frac{1}{N}
\sum_{i=1}^{N}
\left(
\hat y(x_i)-y_i
\right)^2
$$

while the MAE is

$$
\mathrm{MAE}=

\frac{1}{N}
\sum_{i=1}^{N}
\left|
\hat y(x_i)-y_i
\right|
$$

Both metrics quantify discrepancies between theory and experiment. Values closer to zero indicate a better fit to the experimental data.

```python
# ============================================================
# RESIDUAL ANALYSIS
#
# Uses:
#
#   x_exp
#   y_exp
#   Omega_shifted
#   Absorption
#
# already generated by the main model
#
# ============================================================


# ============================================================
# INTERPOLATE MODEL ON EXPERIMENTAL GRID
# ============================================================

Abs_interp = np.interp(

    x_exp,
    Omega_shifted,
    Absorption
)


# ============================================================
# RESIDUAL
#
# Residual = experimental - theoretical
# ============================================================

Residual = 100 * (y_exp - Abs_interp)


# ============================================================
# MODEL QUALITY METRICS
#
# Isotopic model vs experiment
# ============================================================

MSE = np.mean(
    (Abs_interp - y_exp)**2
)

MAE = np.mean(
    np.abs(Abs_interp - y_exp)
)

print()
print("================================")
print("ISOTOPIC MODEL")
print("================================")
print(f"MSE = {MSE:.6f}")
print(f"MAE = {MAE:.6f}")
print("================================")


# ============================================================
# FIGURE WITH RESIDUALS
# ============================================================

fig, (ax1, ax2) = plt.subplots(

    2,
    1,

    figsize=(10,8),

    sharex=True,

    gridspec_kw={
        'height_ratios': [3,1]
    }
)


# ============================================================
# TOP PANEL
#
# Experimental spectrum + theoretical model
# ============================================================

ax1.scatter(

    x_exp,
    y_exp,

    color='black',
    s=25,

    label='Experimental data'
)


ax1.plot(

    Omega_shifted,
    Absorption,

    color='red',
    linewidth=3,

    label='Improved isotopic Gaussian model'
)


# ============================================================
# HYPERFINE LABELS
# ============================================================

ax1.text(
    -4.3,
    0.16,
    r'$^{87}\mathrm{Rb}$' + '\n' + r'$F=1$',
    fontsize=16,
    ha='center'
)

ax1.text(
    -1.76,
    0.36,
    r'$^{85}\mathrm{Rb}$' + '\n' + r'$F=2$',
    fontsize=16,
    ha='center'
)

ax1.text(
    0.24,
    0.88,
    r'$^{85}\mathrm{Rb}$' + '\n' + r'$F=3$',
    fontsize=16,
    ha='center'
)

ax1.text(
    2.6,
    0.48,
    r'$^{87}\mathrm{Rb}$' + '\n' + r'$F=2$',
    fontsize=16,
    ha='center'
)

ax1.text(
    4.25,
    0.95,
    'Pure Rb',
    fontsize=18
)


# ============================================================
# TOP PANEL STYLE
# ============================================================

ax1.set_ylabel(
    'Normalized absorption [a.u.]',
    fontsize=14
)

ax1.grid(alpha=0.3)

ax1.legend(
    fontsize=12,
    loc='upper left'
)


# ============================================================
# BOTTOM PANEL
#
# Residuals
# ============================================================

# Residual trend

ax2.plot(

    x_exp,
    Residual,

    color='black',

    linewidth=1.5,

    zorder=1
)

# Residual points

ax2.scatter(

    x_exp,
    Residual,

    color='dimgray',

    s=25,

    alpha=0.8,

    zorder=2
)


# ============================================================
# MSE / MAE BOX
# ============================================================

ax2.text(

    4.5,
    0.1,

    f"MSE = {MSE:.3e}\nMAE = {MAE:.3e}",

    fontsize=10,

    bbox=dict(
        facecolor='white',
        alpha=0.85
    )
)


# ============================================================
# ZERO LINE
# ============================================================

ax2.axhline(

    0,

    color='gray',

    linestyle='--',

    linewidth=1
)


# ============================================================
# BOTTOM PANEL STYLE
# ============================================================

ax2.set_xlabel(
    'Laser detuning [GHz]',
    fontsize=14
)

ax2.set_ylabel(
    'Difference x 100',
    fontsize=13
)

ax2.grid(alpha=0.3)


# ============================================================
# LAYOUT
# ============================================================

plt.tight_layout()

plt.show()
```

### **INTERACTIVE RUBIDIUM ABSORPTION SPECTRUM**

```python
# ============================================================
# INTERACTIVE RUBIDIUM ABSORPTION SPECTRUM
#
# Improved isotopic Doppler Gaussian model
#
# Includes:
#   - Rb85 and Rb87 separately
#   - Real atomic masses
#   - Natural isotopic abundance
#   - Global spectral shift
#   - Effective peak position corrections
#   - Effective spectral weight corrections
#
# ============================================================
# PHYSICAL CONSTANTS
# ============================================================

# Speed of light [m/s]
c = 299792458                       # [m s^-1]

# Boltzmann constant [J/K]
kB = 1.380649e-23                  # [J K^-1]

# Atomic mass unit [kg]
amu = 1.66053906660e-27            # [kg]


# ============================================================
# RUBIDIUM ISOTOPE MASSES
# ============================================================

# Rb-85 atomic mass [kg]
m85 = 84.911789 * amu              # [kg]

# Rb-87 atomic mass [kg]
m87 = 86.909183 * amu              # [kg]


# ============================================================
# RUBIDIUM D2 TRANSITION
# ============================================================

# D2 transition wavelength [m]
lambda0 = 780.241e-9               # [m]

# Central optical frequency [Hz]
nu0 = c / lambda0                  # [Hz]


# ============================================================
# TEMPERATURE
# ============================================================

# Vapor temperature [K]
T = 293                            # [K]


# ============================================================
# DOPPLER WIDTH FUNCTION
# ============================================================

def doppler_sigma(mass):

    FWHM_Hz = nu0 * np.sqrt(
        (8 * kB * T * np.log(2)) /
        (mass * c**2)
    )

    FWHM_GHz = FWHM_Hz / 1e9

    sigma = FWHM_GHz / (
        2 * np.sqrt(2 * np.log(2))
    )

    return sigma


# ============================================================
# DOPPLER WIDTHS
# ============================================================

sigma85 = doppler_sigma(m85)       # [GHz]
sigma87 = doppler_sigma(m87)       # [GHz]

print("sigma85 =", sigma85, "[GHz]")
print("sigma87 =", sigma87, "[GHz]")


# ============================================================
# NATURAL ISOTOPIC ABUNDANCE
# ============================================================

ab85 = 0.7217
ab87 = 0.2783


# ============================================================
# LOAD EXPERIMENTAL DATA
# ============================================================

data = np.loadtxt(
    "Data_Rb_Gas.csv",
    delimiter=",",
    skiprows=1
)


# ============================================================
# EXPERIMENTAL DATA
# ============================================================

x_exp = data[:,0]                  # [GHz]
y_exp = data[:,1]                  # [arb. units]

# Normalize experimental data
y_exp = y_exp / np.max(y_exp)


# ============================================================
# THEORETICAL HYPERFINE PEAK POSITIONS
# ============================================================

# ------------------------------------------------------------
# Rb87 theoretical centers [GHz]
# ------------------------------------------------------------

centers87_theory = np.array([
    -5.2,
     2.2
])


# ------------------------------------------------------------
# Rb85 theoretical centers [GHz]
# ------------------------------------------------------------

centers85_theory = np.array([
    -1.8,
     1.0
])


# ============================================================
# EFFECTIVE POSITION CORRECTIONS
# ============================================================

# ------------------------------------------------------------
# Rb87 corrections [GHz]
# ------------------------------------------------------------

delta87 = np.array([
    0.54,
    0.00
])


# ------------------------------------------------------------
# Rb85 corrections [GHz]
# ------------------------------------------------------------

delta85 = np.array([
   -0.25,
    0.00
])


# ============================================================
# EFFECTIVE PEAK POSITIONS
# ============================================================

centers87 = centers87_theory + delta87
centers85 = centers85_theory + delta85


# ============================================================
# THEORETICAL SPECTRAL WEIGHTS
# ============================================================

weights87_theory = np.array([
    0.12,
    0.55
])

weights85_theory = np.array([
    0.32,
    1.00
])


# ============================================================
# EFFECTIVE SPECTRAL WEIGHT CORRECTIONS
# ============================================================

delta_weights87 = np.array([
    0.21,
    0.60
])

delta_weights85 = np.array([
    0.00,
    0.00
])


# ============================================================
# EFFECTIVE SPECTRAL WEIGHTS
# ============================================================

weights87 = (
    weights87_theory
    + delta_weights87
)

weights85 = (
    weights85_theory
    + delta_weights85
)


# ============================================================
# GLOBAL SPECTRAL SHIFT
# ============================================================

shift = 0.35                        # [GHz]


# ============================================================
# DETUNING AXIS
# ============================================================

margin = abs(shift)

Omega = np.linspace(
    np.min(x_exp) - margin,
    np.max(x_exp) + margin,
    100000
)


# ============================================================
# BUILD ABSORPTION SPECTRUM
# ============================================================

Absorption = np.zeros_like(Omega)


# ============================================================
# Rb87 CONTRIBUTION
# ============================================================

for W, O0 in zip(weights87, centers87):

    gaussian87 = (
        ab87
        * W
        * np.exp(
            -(Omega - O0)**2 /
            (2 * sigma87**2)
        )
    )

    Absorption += gaussian87


# ============================================================
# Rb85 CONTRIBUTION
# ============================================================

for W, O0 in zip(weights85, centers85):

    gaussian85 = (
        ab85
        * W
        * np.exp(
            -(Omega - O0)**2 /
            (2 * sigma85**2)
        )
    )

    Absorption += gaussian85


# ============================================================
# NORMALIZATION
# ============================================================

Absorption /= np.max(Absorption)


# ============================================================
# APPLY GLOBAL SHIFT
# ============================================================

Omega_shifted = Omega + shift


# ============================================================
# CREATE INTERACTIVE FIGURE
# ============================================================

fig = go.Figure()


# ============================================================
# EXPERIMENTAL DATA
# ============================================================

fig.add_trace(

    go.Scatter(

        x=x_exp,
        y=y_exp,

        mode='markers',

        name='Experimental data',

        marker=dict(
            color='black',
            size=7
        )
    )
)


# ============================================================
# THEORETICAL MODEL
# ============================================================

fig.add_trace(

    go.Scatter(

        x=Omega_shifted,
        y=Absorption,

        mode='lines',

        name='Improved isotopic Gaussian model',

        line=dict(
            color='red',
            width=4
        )
    )
)

# ============================================================
# HYPERFINE LABELS
# ============================================================

# ------------------------------------------------------------
# Left peak : Rb87 F=1
# ------------------------------------------------------------

fig.add_annotation(

    x=-4.3,
    y=0.24,

    text="⁸⁷Rb<br>F = 1",

    showarrow=False,

    font=dict(size=26)
)


# ------------------------------------------------------------
# Second peak : Rb85 F=2
# ------------------------------------------------------------

fig.add_annotation(

    x=-1.76,
    y=0.42,

    text="⁸⁵Rb<br>F = 2",

    showarrow=False,

    font=dict(size=26)
)


# ------------------------------------------------------------
# Main peak : Rb85 F=3
# ------------------------------------------------------------

fig.add_annotation(

    x=0.24,
    y=0.88,

    text="⁸⁵Rb<br>F = 3",

    showarrow=False,

    font=dict(size=26)
)


# ------------------------------------------------------------
# Right peak : Rb87 F=2
# ------------------------------------------------------------

fig.add_annotation(

    x=2.6,
    y=0.545,

    text="⁸⁷Rb<br>F = 2",

    showarrow=False,

    font=dict(size=26)
)


# ============================================================
# PURE RB LABEL
# ============================================================

fig.add_annotation(

    x=4.8,
    y=0.98,

    text="Pure Rb",

    showarrow=False,

    font=dict(size=30)
)


# ============================================================
# LAYOUT
# ============================================================

fig.update_layout(

    title=f"Rubidium absorption spectrum (T = {T} K)",

    width=1100,
    height=650,

    template='plotly_white',

    font=dict(size=18),

    xaxis=dict(

        title='Laser detuning [GHz]',

        range=[-7.2, 7.2],

        showgrid=True,

        zeroline=False
    ),

    yaxis=dict(

        title='Normalized absorption [a.u.]',

        range=[-0.05, 1.05],

        showgrid=True,

        zeroline=False
    ),

    legend=dict(

        x=0.02,
        y=0.98,

        bgcolor='rgba(255,255,255,0.8)',

        bordercolor='black',
        borderwidth=1
    )
)


# ============================================================
# SHOW FIGURE
# ============================================================

fig.show()
```

# **Relation between FWHM and sigma**

Consider a Gaussian profile of the form

$$
G(x)
=
A
\exp\left(
-\frac{(x-x_0)^2}{2\sigma^2}
\right)
$$

where

- $A$ is the maximum amplitude,
- $x_0$ is the distribution center,
- $\sigma$ is the Gaussian standard deviation.

The function reaches its maximum at

$$
x=x_0
$$

with

$$
G(x_0)=A
$$

The FWHM (*full width at half maximum*) is the total width of the curve at half its maximum height.

We therefore seek the points where

$$
G(x)=\frac{A}{2}
$$

Substitution into the Gaussian gives

$$
x-x_0
=
\pm\sigma\sqrt{2\ln2}
$$

The two half-maximum points are

$$
x_1
=
x_0-\sigma\sqrt{2\ln2}
$$

$$
x_2
=
x_0+\sigma\sqrt{2\ln2}
$$

The full width at half maximum is then

$$
\mathrm{FWHM}
=
x_2-x_1
$$

giving

$$
\boxed{
\mathrm{FWHM}
=
2\sqrt{2\ln2}\,\sigma
}
$$

Solving for $\sigma$:

$$
\boxed{
\sigma
=
\frac{\mathrm{FWHM}}
{2\sqrt{2\ln2}}
}
$$

This relation connects the measured Doppler width directly to the Gaussian parameter used in the theoretical model.
