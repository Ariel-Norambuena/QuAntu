# **Theoretical model — Doppler absorption spectrum in rubidium**

## **1. Physical picture**

In a rubidium vapor cell without buffer gas, **thermal Doppler broadening** is the main source of spectral broadening.

Atomic velocities follow the Maxwell–Boltzmann distribution:

$$
f(v)\propto e^{-mv^2/2k_BT}
$$

where

* $m$ is the atomic mass,
* $k_B$ is the Boltzmann constant,
* $T$ is the temperature.

Thermal motion causes each velocity class to experience a different optical frequency through the Doppler effect:

$$
\omega'=\omega-kv
$$

where

* $\omega$ is the laser frequency,
* $k$ is the wave number,
* $v$ is the atomic velocity.

Different velocity classes therefore contribute to absorption at slightly different frequencies, producing Gaussian spectral profiles.

---

# **2. Doppler broadening**

The Doppler width of an optical transition is

$$
\Delta \nu_D =

\nu_0
\sqrt{
\frac{8k_BT\ln2}{mc^2}
}
$$

where

* $\nu_0$ is the transition center frequency,
* $m$ is the atomic mass,
* $c$ is the speed of light.

This is the full width at half maximum (FWHM) of the Doppler profile.

---

# **3. Phenomenological Gaussian model**

The experimental spectrum is modeled as a sum of effective Gaussian profiles:

$$
A(\Omega)
=

\sum_i
A_i
\exp
\left[
-\frac{
(\Omega-\Omega_i)^2
}{
2\sigma^2
}
\right]
$$

where

* $A_i$ is the effective relative amplitude,
* $\Omega_i$ is the effective peak position,
* $\sigma$ is the Gaussian Doppler width,
* $\Omega$ is the laser detuning.

Each Gaussian profile represents an effective spectral contribution associated with experimentally observed hyperfine transitions.

The model is a phenomenological approximation motivated by Doppler-broadening physics.

---

# **4. Relation between FWHM and sigma**

For a Gaussian profile:

$$
\mathrm{FWHM}
=

2\sqrt{2\ln2},\sigma
$$

therefore

$$
\sigma
=

\frac{\mathrm{FWHM}}
{2\sqrt{2\ln2}}
$$

---

# **5. Physical parameters**

## Temperature

The value used is

$$
T = 293\ \mathrm{K}
$$

approximately room temperature.

---

## Optical transition

The rubidium D2 line is used:

$$
\lambda_0 = 780\ \mathrm{nm}
$$

therefore

$$
\nu_0=\frac{c}{\lambda_0}
$$

---

# **6. Calculated Doppler width**

Starting from

$$
\Delta \nu_D =\nu_0
\sqrt{
\frac{8k_BT\ln2}{mc^2}
}
$$

we obtain

$$
\mathrm{FWHM}
\approx
0.50\ \mathrm{GHz}
$$

which gives

$$
\sigma
\approx
0.212\ \mathrm{GHz}
$$

---

# **7. Effective peak positions**

The effective peak positions are

$$
\Omega_i=
(-5.5,,-3.1,,0.0,,1.2)\ \mathrm{GHz}
$$

These positions were fitted phenomenologically to the structure observed in the digitized experimental spectrum.

---

# **8. Effective relative intensities**

The relative amplitudes are

$$
A_i=
(0.10,\ 0.28,\ 1.00,\ 0.45)
$$

These amplitudes are adjustable effective parameters used to reproduce the experimentally observed relative intensities.

---

# **9. Global model shift**

A global shift corrects horizontal offsets associated with digitization of the experimental data:

$$
\Omega \rightarrow \Omega+\Delta
$$

with

$$
\Delta=1.358\ \mathrm{GHz}
$$

---

# **10. Final model expression**

The final model is

$$
A(\Omega)
=

\sum_i
A_i
\exp
\left[
-\frac{
(\Omega-\Omega_i)^2
}{
2\sigma^2
}
\right]
$$

with

$$
\sigma=0.212\ \mathrm{GHz}
$$

followed by the shift

$$
\Omega \rightarrow \Omega+1.358
$$

---

# **11. Physical interpretation**

* Gaussian profiles represent effective spectral contributions associated with experimentally observed hyperfine transitions.

* Peak widths arise mainly from thermal Doppler broadening.

* Positions and amplitudes are effective phenomenological parameters fitted to the experimental data.

* The model reproduces the overall shape of the absorption spectrum.

---

# **12. Model limitations**

The model does not include

* explicit isotopic mixing,
* Voigt profiles,
* homogeneous broadening,
* Zeeman structure,
* optical saturation,
* complex susceptibility,
* optical Bloch equations,
* Clebsch–Gordan coefficients,
* density-matrix dynamics.

The model is therefore a Gaussian phenomenological approximation based on thermal Doppler broadening.

---

```python
# ============================================================
# Rubidium absorption spectrum
#
# Doppler-broadened Gaussian model
# with explicit temperature dependence
#
# ============================================================
# PHYSICAL CONSTANTS
# ============================================================

# Speed of light [m/s]
c = 3.0e8

# Boltzmann constant [J/K]
kB = 1.380649e-23

# Rubidium-85 atomic mass [kg]
m_Rb = 85 * 1.660539e-27


# ============================================================
# RUBIDIUM D2 TRANSITION
# ============================================================

# D2 transition wavelength [m]
lambda0 = 780e-9

# Central optical frequency [Hz]
nu0 = c / lambda0


# ============================================================
# TEMPERATURE
# ============================================================

# Temperature [K]
T = 293


# ============================================================
# DOPPLER FWHM
#
# Δν_D = ν0 sqrt(8 k_B T ln2 / m c²)
# ============================================================

# Doppler linewidth [Hz]
FWHM_Hz = nu0 * np.sqrt(
    (8 * kB * T * np.log(2)) /
    (m_Rb * c**2)
)

# Convert Hz -> GHz
FWHM = FWHM_Hz / 1e9

print("Doppler FWHM =", FWHM, "GHz")


# ============================================================
# GAUSSIAN WIDTH
# ============================================================

# Gaussian sigma [GHz]
sigma = FWHM / (2*np.sqrt(2*np.log(2)))

print("sigma =", sigma, "GHz")


# ============================================================
# LOAD EXPERIMENTAL DATA
# ============================================================

data = np.loadtxt(
    "Data_Rb_Gas.csv",
    delimiter=",",
    skiprows=1
)


# ============================================================
# EXPERIMENTAL DATA
# ============================================================

# Experimental detuning axis [GHz]
x_exp = data[:,0]

# Experimental absorption [arb. units]
y_exp = data[:,1]

# Normalize experimental data
y_exp = y_exp / np.max(y_exp)


# ============================================================
# EFFECTIVE PEAK POSITIONS
# ============================================================

centers = np.array([
    -4.37,
    -1.82,
     1.24,
     2.4
])


# ============================================================
# RELATIVE AMPLITUDES
# ============================================================

amplitudes = np.array([
    0.10,
    0.28,
    1.00,
    0.45
])


# ============================================================
# GLOBAL HORIZONTAL SHIFT
# ============================================================

shift = 0.14


# ============================================================
# THEORETICAL DETUNING AXIS
# ============================================================

# Extra margin to compensate global shift
margin = abs(shift)

# Detuning axis [GHz]
Omega = np.linspace(
    np.min(x_exp) - margin,
    np.max(x_exp) + margin,
    50000
)


# ============================================================
# BUILD GAUSSIAN MODEL
# ============================================================

Absorption = np.zeros_like(Omega)

for A, Omega0 in zip(amplitudes, centers):

    gaussian = A * np.exp(
        -(Omega - Omega0)**2 / (2*sigma**2)
    )

    Absorption += gaussian


# ============================================================
# NORMALIZATION
# ============================================================

# Normalize theoretical absorption
Absorption /= np.max(Absorption)


# ============================================================
# APPLY GLOBAL SHIFT
# ============================================================

Omega_shifted = Omega + shift


# ============================================================
# FIGURE
# ============================================================

plt.figure(figsize=(10,6))


# ============================================================
# EXPERIMENTAL DATA
# ============================================================

plt.scatter(
    x_exp,
    y_exp,
    color='black',
    s=25,
    label='Experimental data'
)


# ============================================================
# THEORETICAL MODEL
# ============================================================

plt.plot(
    Omega_shifted,
    Absorption,
    color='red',
    linewidth=3,
    label='Temperature-dependent Gaussian model'
)

# ============================================================
# HYPERFINE LABELS
# ============================================================

# ------------------------------------------------------------
# Left peak : Rb87 F=1
# ------------------------------------------------------------

plt.text(
    -4.1,
    0.16,
    r'$^{87}\mathrm{Rb}$' + '\n' + r'$F=1$',
    fontsize=16,
    ha='center'
)

plt.text(
    -1.85,
    0.33,
    r'$^{85}\mathrm{Rb}$' + '\n' + r'$F=2$',
    fontsize=16,
    ha='center'
)

plt.text(
    2.5,
    0.88,
    r'$^{85}\mathrm{Rb}$' + '\n' + r'$F=3$',
    fontsize=16,
    ha='center'
)

plt.text(
    2.7,
    0.53,
    r'$^{87}\mathrm{Rb}$' + '\n' + r'$F=2$',
    fontsize=16,
    ha='center'
)

plt.text(
    7.1,
    0.95,
    'Pure Rb',
    fontsize=18
)

# ============================================================
# STYLE
# ============================================================

plt.xlabel(
    'Laser detuning Ω (GHz)',
    fontsize=14
)

plt.ylabel(
    'Normalized absorption',
    fontsize=14
)

# plt.title(
#     f'Rubidium absorption spectrum (T = {T} K)',
#     fontsize=16
# )

plt.grid(alpha=0.3)

plt.legend(
    fontsize=12,
    loc='upper left'
)

plt.tight_layout()

plt.show()
```

### **Difference between experiment and theory**

The lower panel shows the difference between the experimental spectrum and the fitted Gaussian Doppler model.

The theoretical spectrum is interpolated onto the experimental grid, after which the residual is calculated:

$$
R(\Omega)=A_{\mathrm{exp}}-A_{\mathrm{model}}
$$

The difference is multiplied by 100 to highlight small discrepancies between theory and experiment.

Values close to zero indicate good agreement between the model and experimental data.

# **Fit-quality metrics**

The mean squared error (MSE) and mean absolute error (MAE) quantify agreement between the theoretical model and experimental data.

Let

$$
y_i
$$

denote an experimental value and

$$
\hat y(x_i)
$$

the model prediction evaluated at the same point.

The MSE is defined as

$$
\mathrm{MSE}=

\frac{1}{N}
\sum_{i=1}^{N}
\left(
\hat y(x_i)-y_i
\right)^2
$$

while the MAE is

$$
\mathrm{MAE}=

\frac{1}{N}
\sum_{i=1}^{N}
\left|
\hat y(x_i)-y_i
\right|
$$

Both metrics quantify discrepancies between theory and experiment. Values closer to zero indicate a better fit to the experimental data.

```python
# ============================================================
# INTERPOLATE MODEL ON EXPERIMENTAL GRID
# ============================================================

Abs_interp = np.interp(

    x_exp,
    Omega_shifted,
    Absorption
)


# ============================================================
# RESIDUAL
#
# Residual = experimental - theoretical
# ============================================================

Residual = y_exp - Abs_interp


# ============================================================
# MODEL QUALITY METRICS
#
# Gaussian model vs experiment
# ============================================================

MSE = np.mean(
    (Abs_interp - y_exp)**2
)

MAE = np.mean(
    np.abs(Abs_interp - y_exp)
)

print()
print("================================")
print("GAUSSIAN MODEL")
print("================================")
print(f"MSE = {MSE:.6f}")
print(f"MAE = {MAE:.6f}")
print("================================")


# ============================================================
# FIGURE WITH RESIDUALS
# ============================================================

fig, (ax1, ax2) = plt.subplots(

    2,
    1,

    figsize=(10,8),

    sharex=True,

    gridspec_kw={
        'height_ratios': [3,1]
    }
)


# ============================================================
# TOP PANEL
#
# Experimental spectrum + theoretical model
# ============================================================

ax1.scatter(

    x_exp,
    y_exp,

    color='black',
    s=25,

    label='Experimental data'
)


ax1.plot(

    Omega_shifted,
    Absorption,

    color='red',
    linewidth=3,

    label='Temperature-dependent Gaussian model'
)


# ============================================================
# HYPERFINE LABELS
# ============================================================

ax1.text(
    -4.35,
    0.135,
    r'$^{87}\mathrm{Rb}$' + '\n' + r'$F=1$',
    fontsize=16,
    ha='center'
)

ax1.text(
    -1.8,
    0.32,
    r'$^{85}\mathrm{Rb}$' + '\n' + r'$F=2$',
    fontsize=16,
    ha='center'
)

ax1.text(
    2.4,
    0.85,
    r'$^{85}\mathrm{Rb}$' + '\n' + r'$F=3$',
    fontsize=16,
    ha='center'
)

ax1.text(
    2.45,
    0.5,
    r'$^{87}\mathrm{Rb}$' + '\n' + r'$F=2$',
    fontsize=16,
    ha='center'
)


# ============================================================
# CELL LABEL
# ============================================================

ax1.text(
    5.1,
    0.95,
    'Pure Rb',
    fontsize=18
)


# ============================================================
# TOP PANEL STYLE
# ============================================================

ax1.set_ylabel(
    'Normalized absorption',
    fontsize=14
)

ax1.grid(alpha=0.3)

ax1.legend(
    fontsize=12,
    loc='upper left'
)


# ============================================================
# BOTTOM PANEL
#
# Residuals
# ============================================================

# Residual trend

ax2.plot(

    x_exp,
    Residual,

    color='black',

    linewidth=1.5,

    zorder=1
)

# Residual points

ax2.scatter(

    x_exp,
    Residual,

    color='gray',

    s=25,

    alpha=0.8,

    zorder=2
)

# ============================================================
# MSE / MAE BOX
# ============================================================

ax2.text(

    4.2,
    14,

    f"MSE = {MSE:.3e}\nMAE = {MAE:.3e}",

    fontsize=10,

    bbox=dict(
        facecolor='white',
        alpha=0.8
    )
)


# ============================================================
# ZERO LINE
# ============================================================

ax2.axhline(

    0,

    color='gray',

    linestyle='--',

    linewidth=1
)


# ============================================================
# BOTTOM PANEL STYLE
# ============================================================

ax2.set_xlabel(
    'Laser detuning Ω (GHz)',
    fontsize=14
)

ax2.set_ylabel(
    'Difference ×100',
    fontsize=13
)

ax2.grid(alpha=0.3)


# ============================================================
# LAYOUT
# ============================================================

plt.tight_layout()

plt.show()
```

### **INTERACTIVE RUBIDIUM ABSORPTION SPECTRUM**

```python
# ============================================================
# INTERACTIVE Rubidium absorption spectrum
#
# Plotly version
#
# Zoom without losing resolution
# MATLAB-like interaction
#
# ============================================================
# PHYSICAL CONSTANTS
# ============================================================

# Speed of light [m/s]
c = 3.0e8

# Boltzmann constant [J/K]
kB = 1.380649e-23

# Rubidium-85 atomic mass [kg]
m_Rb = 85 * 1.660539e-27


# ============================================================
# RUBIDIUM D2 TRANSITION
# ============================================================

# D2 transition wavelength [m]
lambda0 = 780e-9

# Central optical frequency [Hz]
nu0 = c / lambda0


# ============================================================
# TEMPERATURE
# ============================================================

# Temperature [K]
T = 293


# ============================================================
# DOPPLER FWHM
# ============================================================

# Doppler linewidth [Hz]
FWHM_Hz = nu0 * np.sqrt(
    (8 * kB * T * np.log(2)) /
    (m_Rb * c**2)
)

# Convert Hz -> GHz
FWHM = FWHM_Hz / 1e9

print("Doppler FWHM =", FWHM, "GHz")


# ============================================================
# GAUSSIAN WIDTH
# ============================================================

# Gaussian sigma [GHz]
sigma = FWHM / (2*np.sqrt(2*np.log(2)))

print("sigma =", sigma, "GHz")


# ============================================================
# LOAD EXPERIMENTAL DATA
# ============================================================

data = np.loadtxt(
    "Data_Rb_Gas.csv",
    delimiter=",",
    skiprows=1
)


# ============================================================
# EXPERIMENTAL DATA
# ============================================================

# Experimental detuning axis [GHz]
x_exp = data[:,0]

# Experimental absorption [arb. units]
y_exp = data[:,1]

# Normalize experimental data
y_exp = y_exp / np.max(y_exp)


# ============================================================
# EFFECTIVE PEAK POSITIONS
# ============================================================

# Effective hyperfine peak positions [GHz]
centers = np.array([
    -4.37,
    -1.82,
     1.24,
     2.4
])


# ============================================================
# RELATIVE AMPLITUDES
# ============================================================

# Relative amplitudes [dimensionless]
amplitudes = np.array([
    0.10,
    0.28,
    1.00,
    0.45
])

# ============================================================
# GLOBAL HORIZONTAL SHIFT
# ============================================================

# Horizontal offset correction [GHz]
shift =  0.14


# ============================================================
# THEORETICAL DETUNING AXIS
# ============================================================

# Extra margin to compensate global shift
margin = abs(shift)

# Detuning axis [GHz]
Omega = np.linspace(
    np.min(x_exp) - margin,
    np.max(x_exp) + margin,
    100000
)


# ============================================================
# BUILD GAUSSIAN MODEL
# ============================================================

Absorption = np.zeros_like(Omega)

for A, Omega0 in zip(amplitudes, centers):

    gaussian = A * np.exp(
        -(Omega - Omega0)**2 / (2*sigma**2)
    )

    Absorption += gaussian


# ============================================================
# NORMALIZATION
# ============================================================

Absorption /= np.max(Absorption)


# ============================================================
# APPLY GLOBAL SHIFT
# ============================================================

Omega_shifted = Omega + shift


# ============================================================
# CREATE INTERACTIVE FIGURE
# ============================================================

fig = go.Figure()


# ============================================================
# EXPERIMENTAL DATA
# ============================================================

fig.add_trace(
    go.Scatter(
        x=x_exp,
        y=y_exp,
        mode='markers',
        name='Experimental data',
        marker=dict(
            size=6,
            color='black'
        )
    )
)


# ============================================================
# THEORETICAL MODEL
# ============================================================

fig.add_trace(
    go.Scatter(
        x=Omega_shifted,
        y=Absorption,
        mode='lines',
        name='Temperature-dependent Gaussian model',
        line=dict(
            color='red',
            width=3
        )
    )
)


# ============================================================
# LAYOUT
# ============================================================

fig.update_layout(

    title=f'Rubidium absorption spectrum (T = {T} K)',

    xaxis_title='Laser detuning Ω (GHz)',

    yaxis_title='Normalized absorption',

    template='plotly_white',

    width=1000,
    height=600,

    font=dict(size=16),

    # ========================================================
    # FIX AXIS RANGES
    # ========================================================

    xaxis=dict(
        range=[-7.2, 9.0],
        dtick=2,
        showgrid=True,
        zeroline=False
    ),

    yaxis=dict(
        range=[-0.05, 1.05],
        dtick=0.2,
        showgrid=True,
        zeroline=False
    ),

    # ========================================================
    # LEGEND POSITION
    # ========================================================

    legend=dict(
        x=0.72,
        y=0.98,
        bgcolor='rgba(255,255,255,0.8)'
    )
)


# ============================================================
# INTERACTIVE OPTIONS
# ============================================================

fig.update_xaxes(
    showgrid=True,
    zeroline=False
)

fig.update_yaxes(
    showgrid=True,
    zeroline=False
)


# ============================================================
# SHOW FIGURE
# ============================================================

fig.show()
```

# **Relation between FWHM and sigma**

Consider a Gaussian profile of the form

$$
G(x)
=
A
\exp\left(
-\frac{(x-x_0)^2}{2\sigma^2}
\right)
$$

where

- $A$ is the maximum amplitude,
- $x_0$ is the distribution center,
- $\sigma$ is the Gaussian standard deviation.

The function reaches its maximum at

$$
x=x_0
$$

with

$$
G(x_0)=A
$$

The FWHM (*full width at half maximum*) is the total width of the curve at half its maximum height.

We therefore seek the points where

$$
G(x)=\frac{A}{2}
$$

Substitution into the Gaussian gives

$$
x-x_0
=
\pm\sigma\sqrt{2\ln2}
$$

The two half-maximum points are

$$
x_1
=
x_0-\sigma\sqrt{2\ln2}
$$

$$
x_2
=
x_0+\sigma\sqrt{2\ln2}
$$

The full width at half maximum is then

$$
\mathrm{FWHM}
=
x_2-x_1
$$

giving

$$
\boxed{
\mathrm{FWHM}
=
2\sqrt{2\ln2}\,\sigma
}
$$

Solving for $\sigma$:

$$
\boxed{
\sigma
=
\frac{\mathrm{FWHM}}
{2\sqrt{2\ln2}}
}
$$

This relation connects the measured Doppler width directly to the Gaussian parameter used in the theoretical model.
